```python
#opdracht 2
import numpy as np
import numpy as np
import matplotlib.pyplot as plt

# Beginwaardenhhh
dt = 1                  #s
t_e = 10                   #s
N = int(t_e/dt)+1
t = np.arange(0,t_e+dt,dt) #s

# fietser 1
x_1 = np.zeros(N)          #m     maakt array van lengte N met alleen maar waarde 0.
v_1 = 5                    #m/s

# fietser 2
x_2 = np.zeros(N)          #m     
v_2 = np.zeros(N)          #m/s     
a_2 = .5                   #m/s^2     

# Uitvoeren van de simulatie
for i in range(0,N-1):
    x_1[i+1] = x_1[i] + v_1*dt
    
    x_2[i+1] = x_2[i] + v_2[i]*dt
    v_2[i+1] = v_2[i] + a_2*dt

# Plotten van de gegevens.    
plt.figure()

plt.xlabel('$t$(s)')
plt.ylabel('$x$(m)')

plt.plot(t,x_1,'r.', markersize=1, label='fietser 1')
plt.plot(t,x_2,'b.', markersize=1, label='fietser 2')

plt.show()

```

```python
# Definieer een class voor een deeltje. 
# Een class beschrijft alleen het type eigenschappen dat een ding heeft, bijvoorbeeld: in de beschrijving van de class zeg je:
# - een deeltje heeft een positie.
# - een deeltje heeft een functie die, wanneer aangeroepen, de positie van dat deeltje bijwerkt.
# Merk op dat de class zelf geen deeltje is! 
# Een object dat tot een bepaalde class behoort, wordt een instant van die class genoemd. 
# Je kunt meerdere instants van een class hebben.

class ParticleClass:
    def __init__(self, m, v, r, R):
        self.m = m                         # massa van het deeltje
        self.v = np.array(v, dtype=float)  # snelheids vector
        self.r = np.array(r, dtype=float)  # positie vector
        self.R = np.array(R, dtype=float)  # radius van het deeltje

    def update_position(self, dt):
        """Werk de positie van het deeltje bij op basis van zijn snelheid en tijdstap dt."""
        self.r += self.v * dt


dit_deeltje = ParticleClass(m=0, v=[1, 0], r=[0, 0], R=0.5)
dit_deeltje.update_position(2)
#dat_deeltje = ParticleClass(m=1.0, v=[2.0, 0], r=[5.0, 0.0],R=1.0)

print(dit_deeltje.r)  



```

```python
import numpy as np
import matplotlib.pyplot as plt

class deeltjesklasse:
    def __init__(self, m, v, r, R):
        self.m = m                         # massa van het deeltje
        self.v = np.array(v, dtype=float)  # snelheids vector
        self.r = np.array(r, dtype=float)  # positie vector
        self.R = np.array(R, dtype=float)  # radius van het deeltje
    def update_position(self, dt):
        self.r = self.r + self.v * dt

deeltje_a = deeltjesklasse(m=1.0, v = np.random.uniform(-5,5,1), r = np.random.uniform(-5,5,1), R = 1.0)
deeltje_a.update_position(0.001) 

deeltje_b = deeltjesklasse(m=1.0, v = np.random.uniform(-5,5,1), r = np.random.uniform(-5,5,1), R = 1.0)
deeltje_b.update_position(0.001)

deeltje_c = deeltjesklasse(m=1.0, v = np.random.uniform(-5,5,1), r = np.random.uniform(-5,5,1), R = 1.0)
deeltje_c.update_position(1)

deeltje_d = deeltjesklasse(m=1.0, v = np.random.uniform(-5,5,1), r = np.random.uniform(-5,5,1), R = 1.0)
deeltje_d.update_position(0.001)

deeltje_e = deeltjesklasse(m=1.0, v = np.random.uniform(-5,5,1), r = np.random.uniform(-5,5,1), R = 1.0)
deeltje_e.update_position(0.001)

deeltje_a_array = np.array([])
deeltje_b_array = np.array([])
deeltje_c_array = np.array([])
deeltje_d_array = np.array([])
deeltje_e_array = np.array([])

dt = 0.01
t = 0
t_eind = 5

for i in range(int(t_eind/dt)):
    t = t + dt
    deeltje_a.update_position(dt)
    deeltje_b.update_position(dt)
    deeltje_c.update_position(dt)
    deeltje_d.update_position(dt)
    deeltje_e.update_position(dt)

    deeltje_a_array = np.append(deeltje_a_array, deeltje_a.r)
    deeltje_b_array = np.append(deeltje_b_array, deeltje_b.r)
    deeltje_c_array = np.append(deeltje_c_array, deeltje_c.r)
    deeltje_d_array = np.append(deeltje_d_array, deeltje_d.r)
    deeltje_e_array = np.append(deeltje_e_array, deeltje_e.r)

x = np.linspace(0, t_eind, len(deeltje_a_array))

plt.figure()
plt.plot(x, deeltje_a_array, '.', label='deeltje a')
plt.plot(x, deeltje_b_array, '.', label='deeltje b')
plt.plot(x, deeltje_c_array, '.', label='deeltje c')
plt.plot(x, deeltje_d_array, '.', label='deeltje d')
plt.plot(x, deeltje_e_array, '.', label='deeltje e')
plt.xlabel('tijd (s)')
plt.ylabel('positie op de x as (m)')
plt.legend()
plt.show()
```

```python
# Importeren van libraries
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Maken van de class
class ParticleClass:
    def __init__(self, m, v, a, r, R):
        self.m = m                         
        self.v = np.array(v, dtype=float)
        self.a = a  
        self.r = np.array(r, dtype=float)  
        self.R = R

    def update_position(self):
        self.r += self.v * dt
    def update_velocity(self):
        self.v[1] += self.a * dt

# Simulation parameters
dt = 0.1                                                            # tijd stap
num_steps = 500                                                     # aantal te nemen stappen
particle = ParticleClass(m=1.0, v=[0, 0], a=-9.81, r=[0.0, 0.0], R=1.0)    # het maken van ons deeltje
particlE = ParticleClass(m=1.0, v=[0, 3], a=-9.81, r=[0.0, 0.0], R=1.0)


# Creeer een figuur en de assen
fig, ax = plt.subplots()
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.set_aspect('equal')
ax.set_title("Animatie")
ax.set_xlabel("x")
ax.set_ylabel("y")

# Toon het deeltje als een rode stip
dot, = ax.plot([], [], 'ro', markersize=10); # semicolon to suppress output

# Initialisieren van de functie voor de animatie
def init():
    dot.set_data([], [])
    return dot,

# Updaten van de functie voor elk frame
def update(frame):
    particle.update_velocity()
    particle.update_position()
    particlE.update_velocity()
    particlE.update_position()

    if particlE.r[0] >= 10 and particle.v[0] > 0:
            particlE.v[0] *= -1
    if particlE.r[0] <= -10 and particle.v[0] < 0:
            particlE.v[0] *= -1
    
    if particlE.r[1] >= 10 and particle.v[1] > 0:
            particlE.v[1] *= -1
    if particlE.r[1] <= -10 and particle.v[1] < 0:
            particlE.v[1] *= -1

    if particle.r[0] >= 10 and particle.v[0] > 0:
        particle.v[0] *= -1
    if particle.r[0] <= -10 and particle.v[0] < 0:
        particle.v[0] *= -1

    if particle.r[1] >= 10 and particle.v[1] > 0:
        particle.v[1] *= -1
    if particle.r[1] <= -10 and particle.v[1] < 0:
        particle.v[1] *= -1

    dot.set_data([particle.r[0]], [particle.r[1]])
    dot.set_data([ParticlE.r[0]], [particlE.r[1]])
    return dot,


# Creeer de animatie
ani = FuncAnimation(fig, update, frames=range(200), init_func=init, blit=True, interval=50)

# Omdat we werken met Jup. Notebooks (en niet een .py file)
from IPython.display import HTML
HTML(ani.to_jshtml())

```
